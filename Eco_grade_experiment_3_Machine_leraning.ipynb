```python
def merge_hierarchy(row):

    values = []

    for item in row:

        if pd.notnull(item):

            item = str(item)

            item = item.replace('en:', '')
            item = item.replace('-', ' ')
            item = item.strip()

            if item:
                values.append(item)

    values = list(dict.fromkeys(values))

    return ' '.join(values)
```

```python
df_row['ingredient_hierarchy_text'] = (
    df_row[ingredient_cols]
    .apply(merge_hierarchy, axis=1)
)
df_row['category_hierarchy_text'] = (
    df_row[category_cols]
    .apply(merge_hierarchy, axis=1)
)
```

```python
import re
import unicodedata

def clean_text_lstm(text):

    if pd.isna(text):
        return ''

    text = str(text)

    # unicode normalization
    text = unicodedata.normalize('NFKC', text)

    # remove language tags
    text = re.sub(
        r'\b(?:en|fr|de|pt|it|ru|es|nl):',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    # remove urls
    text = re.sub(
        r'http\S+|www\S+',
        ' ',
        text
    )

    # replace hyphens
    text = text.replace('-', ' ')

    # remove punctuation
    text = re.sub(
        r'[^\w\s]',
        ' ',
        text
    )

    # normalize spaces
    text = re.sub(
        r'\s+',
        ' ',
        text
    ).strip()

    # lowercase
    text = text.lower()

    return text
```

```python
text_cols = [
    'product_name_en',
    'ingredients_text_en',
    'ingredient_hierarchy_text',
    'category_hierarchy_text'
]

for col in text_cols:
    df_nlp[col] = (
        df_nlp[col]
        .fillna('')
        .apply(clean_text_lstm)
    )
```

### Text Vectorization: Bag-of-Words

We will use `CountVectorizer` to convert text data into numerical feature vectors using the Bag-of-Words (BoW) model. This approach counts the occurrences of words within a document, creating a vector where each dimension corresponds to a unique word in the vocabulary.

`max_features` controls the size of the vocabulary, `ngram_range` allows capturing combinations of words (bigrams in this case), and `min_df` filters out words that appear too infrequently.

```python
from sklearn.feature_extraction.text import CountVectorizer

# Bag of Words (BoW) Vectorization
# max_features: Limits the vocabulary size to the most frequent words.
# ngram_range: Includes unigrams and bigrams for better context capture.
# min_df: Ignores terms that appear in less than 2 documents to reduce noise.
bow_vectorizer = CountVectorizer(
    max_features=100000,
    ngram_range=(1,2),
    min_df=2
)

# Fit the vectorizer on the training data and transform both training and test data
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)
```

### Handling Imbalanced Data with SMOTE

Our target variable `ecoscore_grade` might have an imbalanced distribution, which can negatively impact model performance. To address this, we use **SMOTE (Synthetic Minority Over-sampling Technique)**.

SMOTE works by creating synthetic samples from the minority class, helping to balance the class distribution. This prevents models from being biased towards the majority class and improves their ability to classify minority classes correctly.

```python
from imblearn.over_sampling import SMOTE

# Initialize SMOTE with a fixed random state for reproducibility
smote = SMOTE(random_state=42)

# Apply SMOTE to the Bag-of-Words training data
# This will generate synthetic samples for minority classes to balance the dataset
X_train_bow_smote, y_train_bow_smote = smote.fit_resample(
    X_train_bow,
    y_train
)

print("Shape of X_train_bow after SMOTE:", X_train_bow_smote.shape)
print("Distribution of y_train_bow after SMOTE:")
print(pd.Series(y_train_bow_smote).value_counts())
```

### Text Classification Models

This section defines a class `TextClassificationModels` to train and evaluate various machine learning models. As requested, we will focus on a subset of models: **Naive Bayes**, **Random Forest**, and **XGBoost**. These models are well-suited for text classification tasks.

-   **Naive Bayes (MultinomialNB)**: A probabilistic classifier particularly effective for text data, assuming feature independence.
-   **Random Forest (RandomForestClassifier)**: An ensemble learning method that builds multiple decision trees and merges their predictions for improved accuracy and robustness.
-   **XGBoost (XGBClassifier)**: A powerful gradient boosting framework known for its speed and performance, especially on structured and tabular data, which can also be applied to text features.

Each model is trained on the preprocessed training data (`X_train_bow_smote`, `y_train_bow_smote`) and evaluated on the test set (`X_test_bow`, `y_test`) using accuracy and a detailed classification report.

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import classification_report, accuracy_score


class TextClassificationModels:

    def __init__(self):
        self.models = {
            "Naive Bayes": MultinomialNB(),
            "Random Forest": RandomForestClassifier(
                n_estimators=50,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1 # Use all available cores for faster training
            ),
            "XGBoost": XGBClassifier(
                n_estimators=100,
                max_depth=8,
                learning_rate=0.1,
                objective='multi:softmax',
                num_class=7,
                random_state=42
            )
        }

    def train_and_evaluate(self, X_train, X_test, y_train, y_test, label_encoder):

        results = {}

        for name, model in self.models.items():
            print(f"\nTraining {name}...")

            # Fit the model on the training data
            model.fit(X_train, y_train)

            # Make predictions on the test set
            y_pred = model.predict(X_test)

            # Calculate accuracy
            acc = accuracy_score(y_test, y_pred)

            print(f"\n{name} Accuracy: {acc:.4f}")
            # Print detailed classification report
            print(
                classification_report(
                    y_test,
                    y_pred,
                    target_names=label_encoder.classes_
                )
            )

            results[name] = acc

        return results
```

```python
ml_models = TextClassificationModels()

# --- Run models with original Bag-of-Words data (without SMOTE) ---
print("\n--- Training with original Bag-of-Words data ---")
bow_results = ml_models.train_and_evaluate(
    X_train_bow,
    X_test_bow,
    y_train,
    y_test,
    label_encoder
)

# --- Run models with Bag-of-Words data + SMOTE ---
print("\n--- Training with Bag-of-Words data + SMOTE ---")
bow_smote_results = ml_models.train_and_evaluate(
    X_train_bow_smote,
    X_test_bow,
    y_train_bow_smote,
    y_test,
    label_encoder
)

print("\nComparison of results:")
print("Original BoW results:", bow_results)
print("BoW + SMOTE results:", bow_smote_results)
```

### Conclusion on Model Performance

Based on the evaluation of the three models (Naive Bayes, Random Forest, and XGBoost) with and without SMOTE for handling class imbalance, here's a summary of the results:

1.  **Overall Best Performer: XGBoost**
    *   XGBoost consistently achieved the highest accuracy in both scenarios. With the original Bag-of-Words (BoW) data, it scored 0.8266, and with SMOTE-resampled data, its accuracy slightly improved to 0.8308.
    *   This indicates that XGBoost is robust and effective for this text classification task, even with the high-dimensional sparse data from BoW vectorization.

2.  **Random Forest Performance**
    *   Random Forest also performed strongly, with an accuracy of 0.8144 on the original BoW data. Its performance slightly decreased to 0.8130 after applying SMOTE.
    *   While competitive, it was slightly outpaced by XGBoost.

3.  **Naive Bayes Performance**
    *   Naive Bayes showed a decent accuracy of 0.7162 on the original BoW data. However, its performance dropped to 0.6928 when trained on the SMOTE-resampled data.
    *   This suggests that SMOTE, in this particular application, might have introduced noise or synthetic patterns that did not benefit the Naive Bayes model as much, possibly due to its strong assumption of feature independence.

4.  **Impact of SMOTE**
    *   The application of SMOTE aimed to address the class imbalance, particularly for the 'a-plus' category, which has very few samples (36 in the test set). While SMOTE increased the recall for the minority 'a-plus' class for Naive Bayes (from 0.00 to 0.39), it often came at the cost of precision and overall accuracy for Naive Bayes and Random Forest.
    *   For XGBoost, SMOTE led to a minor accuracy improvement, suggesting it might have helped in better learning the decision boundaries across classes, but the macro average metrics across all models show that the `a-plus` class remains challenging to predict accurately.

**Recommendation:**

XGBoost appears to be the most suitable model for this classification task, offering the best balance of accuracy and generalization. While SMOTE's impact on overall accuracy was marginal or negative for Naive Bayes and Random Forest, the slight improvement for XGBoost suggests that further tuning of SMOTE or XGBoost parameters could potentially yield better results, especially for minority classes if their accurate prediction is a critical goal.
