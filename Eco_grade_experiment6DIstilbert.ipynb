### Define Hierarchy Merging Function

This helper function, `merge_hierarchy`, processes a row of hierarchy columns. It cleans and concatenates the non-null hierarchy items into a single string, removing language tags and formatting.

```python
def merge_hierarchy(row):

    values = []

    for item in row:

        if pd.notnull(item):

            item = str(item)

            item = item.replace('en:', '')
            item = item.replace('-', ' ')
            item = item.strip()

            if item:
                values.append(item)

    values = list(dict.fromkeys(values))

    return ' '.join(values)
```

### Apply Hierarchy Merging

The `merge_hierarchy` function is applied to the extracted ingredient and category hierarchy columns to create two new columns: `ingredient_hierarchy_text` and `category_hierarchy_text`. These aggregated text columns will be valuable for NLP tasks.

```python
df_row['ingredient_hierarchy_text'] = (
    df_row[ingredient_cols]
    .apply(merge_hierarchy, axis=1)
)
df_row['category_hierarchy_text'] = (
    df_row[category_cols]
    .apply(merge_hierarchy, axis=1)
)
```

### Define Text Cleaning Function

The `clean_text_lstm` function is designed to preprocess text data for NLP models. It performs several operations: Unicode normalization, removal of language tags, URLs, hyphens, punctuation, normalization of spaces, and lowercasing. This ensures consistency and reduces noise in the text.

```python
print(df_nlp['product_name_en'].sample(10, random_state=1).tolist())
```

```python
(
    (
        df_nlp['product_name_en'].fillna('').str.strip() == ''
    ) &
    (
        df_nlp['ingredients_text_en'].fillna('').str.strip() == ''
    ) &
    (
        df_nlp['ingredient_hierarchy_text'].fillna('').str.strip() == ''
    ) &
    (
        df_nlp['category_hierarchy_text'].fillna('').str.strip() == ''
    )
).sum()
```

```python
import re
import unicodedata

def clean_text_lstm(text):

    if pd.isna(text):
        return ''

    text = str(text)

    # unicode normalization
    text = unicodedata.normalize('NFKC', text)

    # remove language tags
    text = re.sub(
        r'\b(?:en|fr|de|pt|it|ru|es|nl):',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    # remove urls
    text = re.sub(
        r'http\S+|www\S+',
        ' ',
        text
    )

    # replace hyphens
    text = text.replace('-', ' ')

    # remove punctuation
    text = re.sub(
        r'[^\w\s]',
        ' ',
        text
    )

    # normalize spaces
    text = re.sub(
        r'\s+',
        ' ',
        text
    ).strip()

    # lowercase
    text = text.lower()

    return text
```

### Apply Text Cleaning

The `clean_text_lstm` function is applied to all specified text columns in the `df_nlp` DataFrame. This prepares the text data for effective use in NLP models.

```python
text_cols = [
    'product_name_en',
    'ingredients_text_en',
    'ingredient_hierarchy_text',
    'category_hierarchy_text'
]

for col in text_cols:
    df_nlp[col] = (
        df_nlp[col]
        .fillna('')
        .apply(clean_text_lstm)
    )
```

### Import Libraries

Essential libraries for building and training a transformer-based model are imported. This includes `torch` for tensor operations, `numpy` and `pandas` for data handling, `sklearn` for preprocessing and metrics, and `transformers` and `datasets` from Hugging Face for the core model functionalities.

```python
import torch
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report

from transformers import (
    DistilBertTokenizerFast,
    DistilBertForSequenceClassification,
    Trainer,
    TrainingArguments
)

from datasets import Dataset
```

### Create Small Sample for Training (Optional)

Although not strictly necessary for this dataset size given the GPU, a `SAMPLE_SIZE` is defined to potentially limit the training data. For larger datasets, this would be crucial to manage computational resources. Here, it is set to the full training set size.

```python
SAMPLE_SIZE = 137190

X_train_small = X_train[:SAMPLE_SIZE]
y_train_small = y_train[:SAMPLE_SIZE]
```

### Initialize Tokenizer

The `DistilBertTokenizerFast` is loaded. This tokenizer is responsible for converting raw text into numerical input IDs and attention masks that the DistilBERT model can understand. `truncation=True` and `padding=True` ensure that all sequences have a uniform length, and `max_length=128` sets the maximum sequence length.

```python
MODEL_NAME = "distilbert-base-uncased"

tokenizer = DistilBertTokenizerFast.from_pretrained(
    MODEL_NAME
)
```

### Encode Training and Test Data

Both the training and testing `deep_text` data are tokenized using the `DistilBertTokenizerFast`. This generates `input_ids` and `attention_mask` for each text entry, which are the primary inputs for the DistilBERT model. `truncation` and `padding` ensure uniform input size.

```python
train_encodings = tokenizer(
    X_train_small.tolist(),
    truncation=True,
    padding=True,
    max_length=128
)

test_encodings = tokenizer(
    X_test.tolist(),
    truncation=True,
    padding=True,
    max_length=128
)
```

### Create Hugging Face Datasets

The tokenized data (`input_ids`, `attention_mask`) and the corresponding labels are converted into `datasets.Dataset` objects. This format is optimized for use with the Hugging Face `Trainer` API, providing efficient data handling during training.

```python
train_dataset = Dataset.from_dict({
    "input_ids": train_encodings["input_ids"],
    "attention_mask": train_encodings["attention_mask"],
    "labels": y_train_small.tolist()
})

test_dataset = Dataset.from_dict({
    "input_ids": test_encodings["input_ids"],
    "attention_mask": test_encodings["attention_mask"],
    "labels": y_test.tolist()
})
```

### Initialize DistilBERT Model

The `DistilBertForSequenceClassification` model is loaded from a pre-trained checkpoint (`distilbert-base-uncased`). `num_labels=7` specifies that the model should be configured for a 7-class classification task, matching the number of unique `ecoscore_grade` categories. The warning messages indicate that the classification head of the pre-trained model is being re-initialized for this specific task.

```python
model = DistilBertForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=7
)
```

### Configure Training Arguments

`TrainingArguments` define the hyperparameters and settings for the training process. Key parameters include: output directory, evaluation/saving strategy (per epoch), learning rate, batch sizes, number of epochs, weight decay, logging directory, loading the best model at the end, and enabling mixed-precision training (`fp16`).

```python
training_args = TrainingArguments(
    output_dir="./results",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=5,
    weight_decay=0.01,
    logging_dir="./logs",
    load_best_model_at_end=True,
    fp16=True
)
```

### Initialize Trainer

The `Trainer` class from Hugging Face is initialized with the configured model, training arguments, and the training and evaluation datasets. This object encapsulates the entire training and evaluation loop, simplifying the process.

```python
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset
)
```

### Train the Model

This command starts the training process for the DistilBERT model. The `Trainer` handles iterating through epochs, batching data, performing forward and backward passes, and updating model weights according to the specified training arguments.

```python
trainer.train()
```

### Make Predictions

After training, the model makes predictions on the `test_dataset`. The `predictions.predictions` attribute contains the raw logits (unnormalized scores) for each class. `np.argmax` converts these logits into the predicted class labels by selecting the class with the highest score.

```python
predictions = trainer.predict(test_dataset)

y_pred = np.argmax(
    predictions.predictions,
    axis=1
)
```

### Evaluate Model Performance

Finally, the `classification_report` from `sklearn.metrics` is used to evaluate the model's performance on the test set. It provides precision, recall, f1-score, and support for each class (`a` through `f`), as well as overall accuracy and macro/weighted averages. This report gives a comprehensive view of how well the model classifies each `ecoscore_grade`.

```python
 print(
    classification_report(
        y_test,
        y_pred,
        target_names=label_encoder.classes_
    )
)
```
